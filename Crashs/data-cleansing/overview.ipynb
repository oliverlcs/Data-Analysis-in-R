```python
print("Unique Reports:", len(pd.unique(data["Report Number"])))
print("Num of all Reports", len(data["Report Number"]))

# ---
# output:
# Unique Reports: 115360
# Num of all Reports 204688
```

```python
duplicates = data[data.duplicated(subset=['Report Number'], keep=False)]

# Sortieren der Duplikate nach der Berichtsnummer, um sie leichter vergleichen zu können
sorted_duplicates = duplicates.sort_values(by='Report Number')

# Anzeigen der doppelten Einträge
print(sorted_duplicates)
```
